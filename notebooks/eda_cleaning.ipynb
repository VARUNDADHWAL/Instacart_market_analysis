```python
import pandas as pd

# 1. Load the data from the CSV file
df_orders = pd.read_csv("D:\\instacart_market_analysis\\orders.csv")

# 2. Create the new column: is_first_order
df_orders['is_first_order'] = df_orders['days_since_prior_order'].isna().astype(int)

# 3. Replace the null  values with -1 for  visualizations
df_orders['days_since_prior_order'] = df_orders['days_since_prior_order'].fillna(-1)

#verify the changes
print(df_orders.head())

# 5. Save the cleaned DataFrame to a NEW 
# -------------------------------------
# write all new code before this line 
# -------------------------------------
df_orders.to_csv("D:\\instacart_market_analysis\\cleaned_orders.csv", index=False)
```

```python
import pandas as pd

# 1. Load the data from the CSV files
df_prior = pd.read_csv("D:\\instacart_market_analysis\\order_products__prior.csv")
df_train = pd.read_csv("D:\\instacart_market_analysis\\order_products__train.csv")

# 2. Check for missing values in key columns
print("Prior reordered unique values:", df_prior['reordered'].unique())
print("Train reordered unique values:", df_train['reordered'].unique())

# 3. Check for nulls and zeros in add_to_cart_order
print("Prior cart order nulls:", df_prior['add_to_cart_order'].isna().sum())
print("Prior cart order zeros:", (df_prior['add_to_cart_order'] == 0).sum())
print("Train cart order nulls:", df_train['add_to_cart_order'].isna().sum())
print("Train cart order zeros:", (df_train['add_to_cart_order'] == 0).sum())

# 4. Check for duplicates in order_id and product_id combinations
prior_dupes = df_prior.duplicated(subset=['order_id', 'product_id']).sum()
train_dupes = df_train.duplicated(subset=['order_id', 'product_id']).sum()
print("Prior duplicates:", prior_dupes)
print("Train duplicates:", train_dupes)


# -------------------------------------
# write all new code before this line 
# -------------------------------------
# Save cleaned files
df_prior.to_csv("D:\\instacart_market_analysis\\cleaned_order_products__prior.csv", index=False)
df_train.to_csv("D:\\instacart_market_analysis\\cleaned_order_products__train.csv", index=False)
```
